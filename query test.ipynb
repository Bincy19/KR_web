```python
from flask import Flask, render_template
from rdflib import RDFS, Graph, URIRef, Literal
from rdflib.namespace import RDF, OWL, XSD
from rdflib import Graph, Namespace

# Load RDF data from file
rdf_file_path = r'C:/Users/rizwa/Desktop/kr_website/ontology.ttl'
graph = Graph()
graph.parse(rdf_file_path, format='turtle')

# Define the namespaces
onto = Namespace("http://www.kr_paris.com/ontologies/tourism#")
dbo = Namespace("http://dbpedia.org/ontology/")

    

def regions_main():
    # Query to retrieve data
    query_northern = """
        SELECT ?state ?label ?description ?image
        WHERE {
        ?state rdf:type dbo:State ;
                rdfs:label ?label ;
                dbo:description ?description ;
                foaf:depiction ?image .
        FILTER (?state IN (<http://example.org/places/brandenburg>> ,
                            <http://example.org/places/saxony>,
                           <http://example.org/places/saxonyAnhalt> ,
                            <http://example.org/places/berlin> ,
                            <http://example.org/places/thuringia>))
    }
    """

    northen_results = graph.query(query_northern, initNs={"dbo": dbo, "rdf": Namespace("http://www.w3.org/1999/02/22-rdf-syntax-ns#")})

    # Extracting data from the query results and storing in a list of dictionaries
    north_regions_data = [{'n_label': str(row.label), 'n_description': str(row.description), 'n_image': str(row.image)} for row in northen_results]

    # Pass the data to the template
    print(north_regions_data,type(north_regions_data))
regions_main()
```
